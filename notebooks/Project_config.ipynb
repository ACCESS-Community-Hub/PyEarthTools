```python
from pathlib import Path
import os

while True:
    project_home_input = input("Enter the PROJECT_HOME path (or press Enter to use the current working directory): ")
    project_home = Path(project_home_input) if project_home_input else Path.cwd()

    # Check if the provided path is a valid directory
    if project_home.is_dir():
        break  # Exit the loop if the path is valid
    else:
        print(f"The provided path '{project_home}' is not a valid directory. Please try again.")

# Define paths relative to PROJECT_HOME
# Folder to save estimated mean & standard deviation of fields
stats_folder = project_home / "cnn_training/stats"

# Folders used to cache dataset processed by the pipeline
cache_folder = project_home / "cnn_training/cache"

# Folder to save the trained models and other project outputs
default_root_dir = project_home / "cnn_training"

# Ensure that the required directories exist
stats_folder.mkdir(parents=True, exist_ok=True)
cache_folder.mkdir(parents=True, exist_ok=True)
default_root_dir.mkdir(parents=True, exist_ok=True)


# Prompt the user to select an environment
environment = input("Select your environment (NCI, NIWA, MO, Other): ").strip().lower()

if environment == "nci":
    ERA5LOWRES = Path("/g/data/wb00/NCI-Weatherbench/5.625deg")
elif environment == "niwa":
    ERA5LOWRES = Path("/nesi/nobackup/niwa00004/riom/weatherbench/5.625deg")
elif environment == "mo":
    ERA5LOWRES = Path("/data/users/infolab/weatherbench/5.625deg")
elif environment == "other":
    # Allow the user to manually input the path
    era5lowres_input = input("Enter the path to your ERA5 low-resolution data: ").strip()
    ERA5LOWRES = Path(era5lowres_input) if era5lowres_input else None
else:
    ERA5LOWRES = None
    print("Unknown environment. Please set ERA5LOWRES manually if needed.")

# Validate that the ERA5LOWRES path is a directory
if ERA5LOWRES and not ERA5LOWRES.is_dir(): # "and not" Ensures that check works even if the ERA5LOWRES is None
    print(f"The provided path '{ERA5LOWRES}' does not exist. Please set ERA5LOWRES manually if needed.")
    ERA5LOWRES = None

if ERA5LOWRES:
    os.environ["ERA5LOWRES"] = str(ERA5LOWRES)
else:
    print("ERA5LOWRES is not set. Please check your configuration.")

# Print paths for debugging
print(f"Project Home: {project_home}")
print(f"Stats folder: {stats_folder}")
print(f"Cache folder: {cache_folder}")
print(f"Default root directory: {default_root_dir}")
print(f"ERA5LOWRES: {ERA5LOWRES}")

# Show contents of the ERA5LOWRES directory
if ERA5LOWRES:
    print(f"Contents of ERA5LOWRES:")
    !ls -l {ERA5LOWRES}
```
