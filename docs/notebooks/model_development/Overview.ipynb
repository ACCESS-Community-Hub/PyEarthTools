# Model Development with PyEarthTools

To write a new model in PyEarthTools, you will need to use the pyearthtools.training package, and implement the pyearthtools.training.wrapper.ModelWrapper class. One way to think about the concepts here is to imagine that the model is a simple piece of data, and that PyEarthTools is the harness which allows you to interact in a structured way with your chosen machine learning library to train, feed and utilise the model. The model itself is "wrapped" in an API which, although somewhat complex, gives you the tools that you eventually need to do the controlled experiments necessary to achieve a strong modelling result. 

PyEarthTools can be used just for its data preprocessing pipelines, and that's completely fine. However, there are advantages to performing model training and experimentation within the framework also. These advantages include:

- Standard ways to track experiments
- Controlled, repeatable data pre-processing for experiments
- Integration with model scorecards
- Adapt experiments from other compatible codebases to focus on a specific innovation rather than a ground-up rewrite

Tutorial 1, "One-Simplified", gives a simple direct demonstration of training a model in PyEarthTools for people who would like to look ahead at some code. The remaining tutorials step more slowly through the concepts covered in this overview.
