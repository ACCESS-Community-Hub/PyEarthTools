# Accessing CMIP6 data from the Pangeo cloud archive

The [Pangeo CMIP6 cloud archive](https://pangeo-data.github.io/pangeo-cmip6-cloud/accessing_data.html) hosts CMIP6 climate model output as analysis-ready, cloud-optimised (ARCO) Zarr stores on Google Cloud Storage. No account, download or HPC access is needed: PyEarthTools opens the stores lazily, so only the data you use is transferred.

The `PangeoCMIP6` accessor, in the `site_archive_pangeo` package, is derived from the PyEarthTools `ArchiveIndex` class. It looks up each variable in the Pangeo catalogue and opens its Zarr store, so it is used like any other PyEarthTools data accessor.

Install the package from the monorepo with `pip install -e packages/third_party/pangeo_site_archive`.

### Set up imports for demo

```python
import matplotlib.pyplot
import cartopy.crs
import numpy

import pyearthtools.data
import pyearthtools.data.transforms.coordinates
import pyearthtools.pipeline
from site_archive_pangeo import PangeoCMIP6
```

## Discover the available data

The catalogue lists every variable of every model run in the archive. `PangeoCMIP6.catalog()` returns it as a pandas `DataFrame`, so the usual pandas filtering shows what is available. The catalogue is a large CSV file, so the first read takes a moment.

```python
catalog = PangeoCMIP6.catalog()
catalog.head()
```

Each row is one variable of one model run. Here we count, for each experiment, how many models provide monthly near-surface air temperature (`tas` in the `Amon` table).

```python
monthly_tas = catalog[(catalog['variable_id'] == 'tas') & (catalog['table_id'] == 'Amon')]
monthly_tas.groupby('experiment_id')['source_id'].nunique().sort_values(ascending=False).head(10)
```

## Set up the accessor

An accessor selects one model run: the model (`source_id`), the experiment (`experiment_id`), the ensemble member (`member_id`, default `r1i1p1f1`) and the table (`table_id`, default `Amon`, monthly atmosphere). Here we use the `historical` experiment of the `CanESM5` model.

Creating the accessor does not read the catalogue or any data.

```python
index = PangeoCMIP6(['tas'], source_id='CanESM5', experiment_id='historical')
index
```

## Retrieve data

`retrieve` looks up the Zarr store of each variable and selects the requested time. The result is lazy, so the values are only downloaded when they are used.

```python
data = index.retrieve('2000-01')
data
```

## Convert units with a transform and a pipeline

CMIP6 temperatures are in kelvin. In PyEarthTools, operations on data are written as `Transform`s and chained in a `Pipeline`, rather than as one-off array arithmetic, so that they are reusable, recorded and applied the same way to every sample.

Here a `ConvertUnits` transform takes two dictionaries, one of conversion functions and one of the new unit names, both keyed by variable name. It converts the named variables, leaves the others unchanged and updates their `units` attribute.

```python
class ConvertUnits(pyearthtools.data.Transform):
    """
    This transform converts the unit in a dataset.
    """

    def __init__(self, var_convert_dict, var_units):
        self._var_convert_dict = var_convert_dict
        self._var_units = var_units
        if set(var_convert_dict.keys()) != set(var_units.keys()):
            raise ValueError('The unit conversion function and unit names dictionaries must have the same keys.')
        super().__init__()
        self.record_initialisation()

    def apply(self, sample):
        def map_func(da, var_lut):
            try:
                ret_val = var_lut[da.name](da)
            except KeyError:
                ret_val = da
            return ret_val

        ret_sample = sample.map(map_func, var_lut=self._var_convert_dict)
        for var_name, var_unit in self._var_units.items():
            ret_sample[var_name].attrs['units'] = var_unit
        return ret_sample
```

A `Pipeline` takes the data accessor followed by the operations to apply. Indexing the pipeline with a time retrieves the data from the accessor and applies each operation in turn.

```python
var_lut = {'tas': lambda arr1: arr1 - 273.15}
unit_lut = {'tas': 'degC'}

convert_units = ConvertUnits(var_convert_dict=var_lut, var_units=unit_lut)

cmip_pipe = pyearthtools.pipeline.Pipeline(
    index,
    convert_units,
)
cmip_pipe
```

## Visualise the data

Plot the near-surface air temperature of January 2000 on a map, now in degrees Celsius. The model longitudes run from 0 to 360, so the map is centred on 180 degrees.

```python
data_c = cmip_pipe['2000-01']
data_c
```

```python
temperature = data_c['tas'].squeeze()

fig = matplotlib.pyplot.figure(figsize=(10, 5))
ax = matplotlib.pyplot.axes(projection=cartopy.crs.PlateCarree(central_longitude=180))
temperature.plot(
    ax=ax,
    transform=cartopy.crs.PlateCarree(),
    cmap='RdBu_r',
    cbar_kwargs={'label': f"Near-surface air temperature ({temperature.attrs['units']})"},
)
ax.coastlines()
ax.set_title('CanESM5 historical, January 2000');
```

### Supplying the transform to the accessor

Alternatively, give the transform to the accessor with `transforms=`, and every retrieval returns data with the right units.

```python
cmip_index = PangeoCMIP6(
    ['tas'],
    source_id='CanESM5',
    experiment_id='historical',
    transforms=[ConvertUnits(var_convert_dict=var_lut, var_units=unit_lut)],
)
cmip_index.retrieve('2000-01')
```

## Load every time of a variable

A Zarr store holds all times of a variable, so the accessor ignores the requested time when it looks the store up and selects the time afterwards. `filesystem` shows the store that is opened.

```python
index.filesystem('2000-01')
```

`load` opens that store lazily with all of its times. Here we apply the same `ConvertUnits` transform to it, drop the scalar `height` coordinate with the `Drop` transform, and take the area-weighted annual global mean with an `Aggregate` transform in one `TransformCollection`, then plot it for the whole historical run. Only the `tas` data is downloaded, which takes a short while.

```python
drop_height = pyearthtools.data.transforms.coordinates.Drop('height', ignore_missing=True)

def area_weighted_annual_mean(ds, dims):
    weights = numpy.cos(numpy.deg2rad(ds['lat']))
    spatial_mean = ds.weighted(weights).mean(dims, keep_attrs=True)
    return spatial_mean.groupby('time.year').mean(keep_attrs=True)

annual_transforms = pyearthtools.data.transforms.transform.TransformCollection(
    convert_units,
    drop_height,
    pyearthtools.data.transforms.aggregation.Aggregate(area_weighted_annual_mean, reduce_dims=['lat', 'lon']),
)
annual_mean = annual_transforms(index.load(index.filesystem('2000-01')))['tas']

annual_mean.plot()
matplotlib.pyplot.ylabel(f"Global mean near-surface air temperature ({annual_mean.attrs['units']})");
```

## Several variables and grids

Pass a list of variables to retrieve several at once. Each variable is looked up in the catalogue and the newest version is used.

```python
multi_index = PangeoCMIP6(['tas', 'pr'], source_id='CanESM5', experiment_id='historical')
multi_index.retrieve('2000-01')
```

Some variables exist on more than one grid. The accessor then raises an error asking for a `grid_label`, for example `grid_label='gn'` for the native model grid or `grid_label='gr'` for a regridded one.

Scalar coordinates that differ between variables, such as `height` (2 m for `tas`, 10 m for `uas`), are dropped when the variables are combined.
